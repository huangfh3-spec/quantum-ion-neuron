```python
from pathlib import Path

import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np

# Fig. S10: J1 coupling waveform used for the fast dissipative channel.
# The values are the final J1 sequence from the original simulation notebook.
TIME_STEP_US = 4
TIME_US = np.arange(0, 160 + TIME_STEP_US, TIME_STEP_US, dtype=float)
J1_MHZ = np.array([0.0, 0.0, 0.07672047830920511, 0.13231679463974813, 0.1859078536139534, 0.23789023884571897, 0.28808878627412293, 0.33620856096787055, 0.38192530590810975, 0.42491718618019214, 0.4648814555242785, 0.5015457081769418, 0.5346762761409519, 0.5640844360794203, 0.589630562353904, 0.6112262689181905, 0.6288346266711229, 0.6424686836175122, 0.6521885653575751, 0.6580975786769886, 0.6603377221831965, 0.6590850511224283, 0.6545452844240113, 0.6469500048604723, 0.6365537279804426, 0.6236320517344519, 0.6084810290780328, 0.5914178334510788, 0.5727827297778861, 0.5529422404516755, 0.5322932840438724, 0.5112677690793862, 0.490336771284797, 0.47001279865829837, 0.450847848291655, 0.4334240887403725, 0.41833355709392217, 0.40614403150268297, 0.397351300904816, 0.3923236417481773, 0.39125069010122726], dtype=float)

if len(TIME_US) != len(J1_MHZ):
    raise ValueError("TIME_US and J1_MHZ must have the same length.")

mpl.rcParams.update({
    "font.family": "DejaVu Sans",
    "mathtext.fontset": "dejavusans",
    "axes.linewidth": 0.9,
    "pdf.fonttype": 42,
    "ps.fonttype": 42,
    "svg.fonttype": "none",
})

fig, ax = plt.subplots(figsize=(6.8, 2.8), dpi=300)
ax.step(TIME_US, J1_MHZ, where="post", color="#4A07C7", linewidth=1.8)

ax.set_xlim(0, 160)
ax.set_ylim(-0.02, 0.70)
ax.set_xticks(np.arange(0, 161, 25))
ax.set_yticks([0.0, 0.2, 0.4, 0.6])
ax.set_xlabel("Time (μs)")
ax.set_ylabel(r"$J_1$ (MHz)")

ax.tick_params(axis="both", which="major", direction="out", length=4, width=0.9)
for spine in ax.spines.values():
    spine.set_linewidth(0.9)

fig.tight_layout(pad=0.4)
for suffix in ("pdf", "svg"):
    fig.savefig(Path(f"FigS10.{suffix}"), bbox_inches="tight")
plt.show()
```
